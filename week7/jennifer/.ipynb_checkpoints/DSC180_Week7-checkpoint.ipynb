### Task 2: First Gradient to Check Against IPCC

```python
# imports

import os, glob, pickle
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
import jax, jax.numpy as jnp
```

```python
# import and load the jax emulator

jax.config.update("jax_enable_x64", False)
print(jax.devices())

# Point this at Cheryl's results folder
CHERYL_DIR = "../cheryl/results"   # adjust if your path differs
DATA_ROOT = "/home/jtc006/dsc180-D03/week7/jennifer"

with open(f"{CHERYL_DIR}/emulator.pkl", "rb") as f:
    bundle = pickle.load(f)

STACKED = jax.tree_util.tree_map(jnp.asarray, bundle["params"])
PRE_FROZEN = {k: jnp.asarray(v) for k, v in bundle["pre"].items()}
NLAT, NLON = bundle["nlat"], bundle["nlon"]
lat, lon = bundle["lat"], bundle["lon"]

area_w_np = np.cos(np.deg2rad(lat))[:, None] * np.ones((1, NLON))
area_w = jnp.asarray(area_w_np / area_w_np.mean(), dtype=jnp.float32)
print("Loaded emulator. lat/lon:", NLAT, NLON)
```

```python
# rebuild the forward functions

def features(co2, ch4, so2, bc, pre):
    T = co2.shape[0]
    so2_f = (so2.reshape(T, -1) - pre["so2_mean"]) @ pre["so2_pcs"].T
    bc_f  = (bc.reshape(T, -1)  - pre["bc_mean"])  @ pre["bc_pcs"].T
    return jnp.concatenate([co2[:, None], ch4[:, None], so2_f, bc_f], axis=1)

def mlp(params, x):
    *layers, skip = params
    h = x
    for layer in layers[:-1]:
        h = jnp.tanh(h @ layer["W"] + layer["b"])
    return h @ layers[-1]["W"] + layers[-1]["b"] + x @ skip["W"] + skip["b"]

def predict(params, co2, ch4, so2, bc, pre):
    x = (features(co2, ch4, so2, bc, pre) - pre["f_mean"]) / pre["f_std"]
    coeffs = mlp(params, x) * pre["c_std"]
    maps = pre["tas_mean"] + coeffs @ pre["tas_pcs"]
    return maps.reshape(co2.shape[0], NLAT, NLON)

# STACKED has an extra leading "seed" axis on every leaf
def one_seed(i):
    return jax.tree_util.tree_map(lambda a: a[i], STACKED)

n_seeds = STACKED[0]["W"].shape[0]

@jax.jit
def emulate(co2, ch4, so2, bc):
    """ClimateBench inputs for one scenario -> tas, shape (years, lat, lon)."""
    per_seed = jax.vmap(lambda p: predict(p, co2, ch4, so2, bc, PRE_FROZEN))(STACKED)
    return per_seed.mean(0)
```

```python
# load ssp245 inputs and sanity check
def find(name):
    hits = glob.glob(os.path.join(DATA_ROOT, "**", name), recursive=True)
    if not hits:
        raise FileNotFoundError(f"{name} not found under {DATA_ROOT}")
    return hits[0]

X = xr.open_dataset(find("inputs_ssp245.nc"))
years = X.time.values
co2 = jnp.asarray(X["CO2"].values.astype(np.float32))
ch4 = jnp.asarray(X["CH4"].values.astype(np.float32))
so2 = jnp.asarray(X["SO2"].transpose("time", ...).values.astype(np.float32))
bc  = jnp.asarray(X["BC"].transpose("time", ...).values.astype(np.float32))
print(f"ssp245 years {years[0]}-{years[-1]}, {len(years)} total")

# Quick sanity check: emulator should produce reasonable global-mean tas
tas_pred = emulate(co2, ch4, so2, bc)
gm = (tas_pred * area_w).mean((-2, -1))
print(f"global-mean tas range: {float(gm.min()):.2f} to {float(gm.max()):.2f} K")
```

```python
# Task 2.1: Gradient of J wrt CO₂, convert to effect of extra unit emitted

# J = global-mean tas averaged over 2081-2100 under ssp245
START, END = 2081, 2100
sel = (years >= START) & (years <= END)
sel_j = jnp.asarray(sel)

def J(co2_vec):
    tas = emulate(co2_vec, ch4, so2, bc)
    gm_all = (tas * area_w).mean((-2, -1))
    # average over 2081-2100
    return jnp.sum(gm_all * sel_j) / jnp.sum(sel_j)

J0 = float(J(co2))
grad_cum = jax.grad(J)(co2)         # dJ / d(cumulative CO2 in year t)
grad_cum_np = np.asarray(grad_cum)
print(f"J (2081-2100 mean tas, ssp245) = {J0:.3f} K")
print(f"grad wrt cumulative CO2, first 5 years: {grad_cum_np[:5]}")
print(f"grad wrt cumulative CO2, last 5 years:  {grad_cum_np[-5:]}")

# Convert to "effect of emitting one extra unit in year t"
# One extra unit emitted in year t raises cumulative total in every year s >= t,
# so dJ/d(annual emissions in year t) = sum_{s>=t} dJ/d(cumulative[s])
dJ_dannual = np.cumsum(grad_cum_np[::-1])[::-1]
print(f"\nEffect of extra unit in year t, first 5 years: {dJ_dannual[:5]}")
print(f"Effect of extra unit in year t, last 5 years:  {dJ_dannual[-5:]}")

plt.figure(figsize=(8, 4))
plt.plot(years, dJ_dannual)
plt.xlabel("year of emission")
plt.ylabel("dJ / d(annual CO2 in year t)")
plt.title("Effect on 2081-2100 warming of one extra unit of CO2 emitted in year t")
plt.axhline(0, color="k", lw=0.5)
plt.grid(alpha=0.3)
plt.show()
```

```python
# Task 2.2: Estimate TCRE and compare to IPCC

# TCRE = warming per unit of cumulative CO2 emitted
# First check the units of the CO2 input
print(f"CO2 input range on ssp245: {float(co2.min()):.1f} to {float(co2.max()):.1f}")
print(f"CO2 units from file: {X['CO2'].attrs.get('units', 'not specified')}")

# ClimateBench typically gives cumulative CO2 in Gt CO2 (or sometimes GtC — check!)
# IPCC best estimate: 0.45 K per 1000 GtCO2 (range 0.27-0.63)

# Simplest TCRE estimate: the effect of an early-year emission,
# since late-year emissions have less time to act
tcre_per_unit = float(dJ_dannual[0])   # K per unit of CO2 emitted in year[0]
print(f"\ndJ/d(annual CO2) in first year: {tcre_per_unit:.6e} K per unit CO2")

# If units are Gt CO2, multiply by 1000 to get K per 1000 GtCO2
tcre_per_1000Gt = tcre_per_unit * 1000
print(f"TCRE estimate: {tcre_per_1000Gt:.3f} K per 1000 GtCO2")
print(f"IPCC best estimate: 0.45 K per 1000 GtCO2 (likely range 0.27-0.63)")
```

```python
# Graphing cumulative CO2 emissions overtime based on ssp245 data
plt.figure(figsize=(8, 4))
plt.plot(years, np.asarray(co2), 'k-', lw=2)
plt.xlabel("year"); plt.ylabel("cumulative CO2 (Gt)")
plt.title("ssp245 cumulative CO2 emissions")
plt.grid(alpha=0.3); plt.show()
```

```python
# Task 2.3: Finite-difference check — match what grad_cum means
# grad_cum[t] = dJ/d(cumulative CO2 at year t), so FD perturbs cumulative at one year

# Pick a year in the late part of the record where the gradient is nonzero
# Find a year where the gradient is nonzero
nonzero_years = np.where(np.abs(grad_cum_np) > 1e-10)[0]
print(f"Years where grad_cum is nonzero: {years[nonzero_years]}")

t_check = nonzero_years[-1]   # use the last nonzero year
h = 1.0

co2_plus  = co2.at[t_check].add(h)
co2_minus = co2.at[t_check].add(-h)

J_plus  = float(J(co2_plus))
J_minus = float(J(co2_minus))
fd = (J_plus - J_minus) / (2 * h)

print(f"\nFinite difference in year {int(years[t_check])}: {fd:.6e} K per unit CO2")
print(f"Autodiff gradient (cumulative) in same year: {float(grad_cum[t_check]):.6e} K per unit CO2")
if abs(float(grad_cum[t_check])) > 1e-12:
    print(f"Ratio: {fd / float(grad_cum[t_check]):.4f}  (should be ~1.0)")
```

**TCRE = Transient Climate Response to cumulative Emissions.**

It measures how much global-mean surface temperature rises per unit of cumulative CO₂ emitted — so units of K per 1000 GtCO₂ (or sometimes K per TtC, trillion tons of carbon). Your emulator estimated 0.504 K per 1000 GtCO₂.

```python
labels = ["IPCC low", "IPCC central", "Your emulator", "IPCC high"]
vals = [0.27, 0.45, 0.504, 0.63]
colors = ["lightgray", "gray", "steelblue", "lightgray"]
plt.figure(figsize=(7, 4))
bars = plt.bar(labels, vals, color=colors)
plt.ylabel("TCRE (K per 1000 GtCO2)")
plt.title("TCRE: emulator vs IPCC AR6")
plt.axhspan(0.27, 0.63, alpha=0.15, color='green', label="IPCC likely range")
for bar, v in zip(bars, vals):
    plt.text(bar.get_x() + bar.get_width()/2, v + 0.01, f"{v:.3f}", ha='center')
plt.legend(); plt.grid(alpha=0.3, axis='y'); plt.show()
```
